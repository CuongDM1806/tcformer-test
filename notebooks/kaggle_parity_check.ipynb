# Parity check: `main` against the code that produced the paper's results

For each run reported in the paper, the same fold (target subject 1) is built twice on CPU, once with the
development branch that produced the result and once with `main`, and compared: the four data splits,
the initial weights, the weights after a few training steps, and the test logits after Stage-2 adaptation.
`SAME` in every column means both code bases compute the same thing with the same random numbers.

No GPU needed, Internet on. For OpenBMI labels attach `openbmi-dataset-cuong`.

```python
LABELS = None           # None = every BCI IV-2a/2b and Zhou2016 run; or e.g. ['bcic2a/full', 'openbmi/no_ra']
STEPS = 6               # training steps compared
BRANCH = 'main'
```

```python
import subprocess, sys
from pathlib import Path

REPO = Path('/tmp/ssm-dan')                 # outside /kaggle/working: not part of the output
if not (REPO / '.git').exists():
    subprocess.run(['git', 'clone', '--depth', '1', '--branch', BRANCH,
                    'https://github.com/CuongDM1806/tcformer-test.git', str(REPO)], check=True)
subprocess.run(['git', '-C', str(REPO), 'log', '-1', '--oneline'], check=True)
sys.path.insert(0, str(REPO))
from ssm_dan import kaggle

kaggle.parity(LABELS, steps=STEPS)
```
